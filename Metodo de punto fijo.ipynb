# **Método de punto fijo**

La ecuación $x^3 + 4x^2 - 10 = 0$ tiene una raíz única en $[1,2]$.

Existen muchas formas de cambiar la ecuación a la forma de punto fijo $x = g(x)$ mediante una simple manipulación algebraica.

Por ejemplo, para obtener la función $g$ descrita en la parte c), podemos manipular la ecuación:

$$
x^3 + 4x^2 - 10 = 0
$$

Como sigue:

$$
4x^2 = 10 - x^3
$$

Por lo que:

$$
x^2 = \frac{1}{4}(10-x^3)
$$

Y finalmente:

$$
x = \pm \frac{1}{2}(10-x^3)^{1/2}
$$

Para obtener una solución positiva, se selecciona $g_3(x)$.

No es importante derivar las funciones que se muestran aquí, pero debemos verificar que el punto fijo de cada una es en realidad una solución para la ecuación original:

$$
x^3 + 4x^2 - 10 = 0
$$



```python
#Importamos la librería numpy para funciones matemáticas y matplotlib que permite generar gráficas.
import numpy as np
import matplotlib.pyplot as plt
```

```python
#función original
def f(x):
    return x**3+4*x**2-10
```

```python
#Generar gráfica
xx=np.linspace(1,2)
plt.plot(xx,f(xx))
plt.axhline(y=0, color='green')
plt.axvline(x=0, color='red')
```

```python
#Esta es la función despejada
def g(x):
    return np.sqrt((10-x**3)/4)
```

```python
g(1.5)
```

```python
#Método de punto fijo
def pfijo(p0,i):

   p0=1.5
   lista=[]
   error=[]

   for i in range(10):
      p1=g(p0)
      lista.append(p1)
      e=np.abs(p1-p0)
      error.append(e)
      p0=p1
   return lista, error
```
